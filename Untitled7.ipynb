```python
!pip install ucimlrepo

import ucimlrepo
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from ucimlrepo import fetch_ucirepo
#Imports for neural network
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Dropout
from tensorflow.keras.models import Sequential

import gradio as gr
# fetch dataset
chronic_kidney_disease = fetch_ucirepo(id=336)


# data (as pandas dataframes)
X = chronic_kidney_disease.data.features
y = chronic_kidney_disease.data.targets
for col in X.columns:
    print(col)
for col in X.columns:
    if X[col].dtype == 'object':
        X[col] = X[col].astype(str).str.strip()
        X[col] = LabelEncoder().fit_transform(X[col])
X.fillna(X.median(numeric_only = True))  #If any column value is missing/blank, fills it in with the avg value for the col


le = LabelEncoder() #Converts labels like column names and such into 0's and 1's
y = le.fit_transform(y.values.ravel())#Converts y values to 1-dimensional and uses the label encoder on them
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_CNN =  X_scaled.reshape(X_scaled.shape[0], X_scaled.shape[1],1) #CNN Requires X, so found a method to convert into 3D numpy

# metadata
print(chronic_kidney_disease.metadata)

# variable information
print(chronic_kidney_disease.variables)

for col in X.columns:
    print("hi!")
'''model = Sequential([
    Conv1D(filters = 32, kernel_size = 3, activation = 'relu', input_shape = (X_CNN.shape[1], 1)), #32 means learns 32 patterns
    MaxPooling1D(pool_size = 2), #Creates ppoling which helps limit the unimportant data and keeps only the import and significant data, pool_size makes it a 2X2 size for the pool when it goes onto the input data array
    Conv1D(filters=64, kernel_size=3, activation='relu'),   #Anotehr Conv1D is made so that another kernel filter is made, this helps see larger patterns bcuz there is 2 now
    MaxPooling1D(pool_size = 2), #Anotehr pooling process to make sure again that majority
    Conv1D(filters = 128, kernel_size = 3, activation = 'relu'),
    MaxPooling1D(pool_size=2),
    Flatten(), #Makes sure all 2D is converted to 1D
    Dense(32,activation = 'relu'), #32 neurons, common use tbh
    Dense(1, activation = 'sigmoid')

])'''
model = Sequential([
    Dense(64, activation='relu', input_shape=(X.shape[1],)),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,y,random_state = 42, test_size = 0.2)

model.compile( #Compiles into a format which can be ready to be used
    optimizer='adam',       #Adjusts weight effeciently!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!
    loss='binary_crossentropy',    #Works in 2-class problems like ours
    metrics=['accuracy'] #Tracks our accuracy

)
model.fit(
    X_train, y_train,
    epochs=30,
    batch_size=16,
    validation_split=0.2
)
loss, acc = model.evaluate(X_test, y_test)
print(acc)
inputs = [
    gr.Number(label=col) for col in X.columns
]
column_mapping = {
    "age": "Age",
    "bp": "Blood_Pressure",
    "sg": "Specific_Gravity",
    "al": "Albumin",
    "su": "Sugar",
    "rbc": "Red_Blood_Cells",
    "pc": "Pus_Cell",
    "pcc": "Pus_Cell_Clumps",
    "ba": "Bacteria",
    "bgr": "Blood_Glucose_Random",
    "bu": "Blood_Urea",
    "sc": "Serum_Creatinine",
    "sod": "Sodium",
    "pot": "Potassium",
    "hemo": "Hemoglobin",
    "pcv": "Packed_Cell_Volume",
    "wc": "White_Blood_Cell_Count",
    "rc": "Red_Blood_Cell_Count",
    "htn": "Hypertension",
    "dm": "Diabetes_Mellitus",
    "cad": "Coronary_Artery_Disease",
    "appet": "Appetite",
    "pe": "Pedal_Edema",
    "ane": "Anemia"
}
X = X.rename(columns = column_mapping)

def predict_ckd(*inputs):
    input_array = np.array(inputs).reshape(1, -1)
    input_scaled = scaler.transform(input_array)
    prediction = model.predict(input_scaled)[0][0]

    if prediction > 0.5:
        return "Chronic Kidney Disease Detected"
    else:
        return "No Chronic Kidney Disease"
# Create interface
interface = gr.Interface(
    fn=predict_ckd,  # prediction function
    inputs=inputs,   # input fields
    outputs="text",  # output type
    title="Chronic Kidney Disease Predictor",
    description="Enter the patients medical information to predict Chronic Kidney Disease",
    theme=gr.themes.Soft(
        primary_hue="cyan",
        secondary_hue="cyan"
    )



)
interface.launch()
```
